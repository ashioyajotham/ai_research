## 🧠 Step 3 — Load Gemma 4 E4B in 4-bit

**Why E4B?** It's the largest Gemma 4 variant that fits comfortably in a T4's 16GB VRAM at 4-bit precision.
The "E" stands for *effective* parameters — it uses Per-Layer Embeddings (PLE) to punch above its weight class.

````python
import torch
import time
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

MODEL_ID = "google/gemma-4-e4b-it"

from transformers import CONFIG_MAPPING
if 'gemma4' not in CONFIG_MAPPING:
    raise RuntimeError(
        "gemma4 not in CONFIG_MAPPING.\n"
        "Go to Runtime → Restart session, then re-run from Step 1."
    )

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)

print(f"⏳ Loading {MODEL_ID}...")
t0 = time.time()

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config,
    device_map="auto",
    dtype=torch.bfloat16,          # torch_dtype is deprecated in new transformers
    attn_implementation="eager",   # safe fallback for Gemma4's hybrid attention
)

load_time = time.time() - t0
vram_used = torch.cuda.memory_allocated() / 1e9
vram_total = torch.cuda.get_device_properties(0).total_memory / 1e9

# hf_device_map only exists on standard CausalLM — Gemma4ForConditionalGeneration
# is a multimodal class and doesn't set it. Fall back gracefully.
device_info = getattr(model, 'hf_device_map', None) or str(next(model.parameters()).device)

print(f"✅ Model loaded in {load_time:.1f}s")
print(f"📦 VRAM used: {vram_used:.2f} GB / {vram_total:.0f} GB")
print(f"🔧 Device: {device_info}")
print(f"🏗️  Model class: {type(model).__name__}")
````

## 🔧 Step 4 — The Agent Harness

A lightweight harness that:
- Wraps generation with timing + token counting
- Extracts code blocks from the response
- Writes files to disk so we can actually *run* the generated code
- Tracks intervention count (how many times we had to re-prompt)

````python
import re
import os
import subprocess
from dataclasses import dataclass
from typing import Optional
from IPython.display import display, Markdown

@dataclass
class RunMetrics:
    """Scorecard for the demo comparison."""
    model_name: str
    steps: int = 0
    interventions: int = 0
    total_time_s: float = 0.0
    errors_self_corrected: int = 0
    tests_passed: Optional[bool] = None
    tokens_generated: int = 0

    def display(self):
        status = "✅ PASS" if self.tests_passed else ("❌ FAIL" if self.tests_passed is False else "⏳ N/A")
        display(Markdown(f"""
### 📊 Scorecard — `{self.model_name}`
| Metric | Result |
|---|---|
| Steps to working code | {self.steps} |
| Human interventions | {self.interventions} |
| Time to completion | {self.total_time_s:.1f}s |
| Errors self-corrected | {self.errors_self_corrected} |
| Tokens generated | {self.tokens_generated:,} |
| Test suite result | {status} |
"""))


def generate(
    prompt: str,
    system: str = "You are an expert Python software engineer. Write complete, working code.",
    max_new_tokens: int = 2048,
    temperature: float = 0.2,
) -> tuple[str, float, int]:
    """Run a single generation pass. Returns (response_text, elapsed_s, n_tokens)."""
    messages = [
        {"role": "system", "content": system},
        {"role": "user",   "content": prompt},
    ]

    # apply_chat_template may return a raw tensor OR a BatchEncoding dict
    # depending on transformers version + model type. Normalise explicitly.
    raw = tokenizer.apply_chat_template(
        messages,
        return_tensors="pt",
        add_generation_prompt=True,
    )

    if hasattr(raw, "input_ids"):
        # BatchEncoding — extract tensors and move to device
        input_ids      = raw.input_ids.to(model.device)
        attention_mask = raw.attention_mask.to(model.device)
        gen_kwargs = {"input_ids": input_ids, "attention_mask": attention_mask}
    else:
        # Plain tensor
        input_ids  = raw.to(model.device)
        gen_kwargs = {"input_ids": input_ids}

    n_input = input_ids.shape[-1]

    t0 = time.time()
    with torch.inference_mode():
        output_ids = model.generate(
            **gen_kwargs,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            do_sample=temperature > 0,
            pad_token_id=tokenizer.eos_token_id,
        )
    elapsed = time.time() - t0

    new_tokens = output_ids[0][n_input:]
    response = tokenizer.decode(new_tokens, skip_special_tokens=True)
    return response, elapsed, len(new_tokens)


def extract_code_blocks(text: str) -> dict[str, str]:
    """
    Extract fenced code blocks. Infers filename from first-line comment
    e.g. '# main.py' — otherwise names them block_1.py, block_2.py etc.
    """
    pattern = r"```(?:\w+)?\n(.*?)```"
    blocks = {}
    for i, m in enumerate(re.finditer(pattern, text, re.DOTALL)):
        code = m.group(1).strip()
        first_line = code.split("\n")[0]
        fname_match = re.match(r"#\s*([\w./]+\.py)", first_line)
        key = fname_match.group(1) if fname_match else f"block_{i+1}.py"
        blocks[key] = code
    return blocks


def write_files(blocks: dict[str, str], output_dir: str = "/tmp/gemma_output") -> list[str]:
    """Write extracted code blocks to disk. Auto-renames test files if needed."""
    os.makedirs(output_dir, exist_ok=True)
    written = []
    for fname, code in blocks.items():
        base = os.path.basename(fname)

        # Pytest only discovers test_*.py — rename common variants automatically
        test_aliases = {"tests.py", "api_test.py", "test.py", "tests_main.py"}
        has_test_fn  = "def test_" in code
        not_named    = not base.startswith("test_")

        if has_test_fn and not_named:
            # Check if we already have a test file being written this round
            existing_test = next((b for b in blocks if os.path.basename(b).startswith("test_")), None)
            if existing_test is None or base in test_aliases:
                base = "test_main.py"
                print(f"  ⚠️  Renamed '{fname}' → 'test_main.py' (pytest discovery fix)")

        path = os.path.join(output_dir, base)
        with open(path, "w") as f:
            f.write(code)
        written.append(path)
        print(f"  📄 Written: {path}")
    return written


def run_tests(test_dir: str = "/tmp/gemma_output") -> tuple[bool, str]:
    """Run pytest. Returns (passed, output). Diagnoses empty collection."""
    files = os.listdir(test_dir) if os.path.isdir(test_dir) else []
    test_files = [f for f in files if f.startswith("test_") and f.endswith(".py")]

    if not test_files:
        # Diagnose: list what IS there so we can fix it
        diag = f"⚠️  pytest collected 0 items — no test_*.py files found.\n"
        diag += f"   Files in {test_dir}: {files}\n"
        for f in files:
            path = os.path.join(test_dir, f)
            snippet = open(path).read()[:300]
            has_tests = "def test_" in snippet
            diag += f"   {f}: has 'def test_'={has_tests}\n"
        return False, diag

    result = subprocess.run(
        ["python", "-m", "pytest", test_dir, "-v", "--tb=short", "--no-header"],
        capture_output=True, text=True, cwd=test_dir
    )
    return result.returncode == 0, result.stdout + result.stderr


print("✅ Harness ready")
````

## 🚀 Step 6 — Run Gemma 4 on the Task

⏱️ *This will take ~2–4 minutes on T4 at 4-bit. Watch the token count.*

````python
metrics = RunMetrics(model_name="Gemma 4 E4B (4-bit, T4)")
OUTPUT_DIR = "/tmp/gemma_output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# conftest.py — adds output dir to sys.path so 'from main import ...' works
with open(f"{OUTPUT_DIR}/conftest.py", "w") as f:
    f.write("import sys, os\nsys.path.insert(0, os.path.dirname(__file__))\n")

t_total_start = time.time()

# ── PASS 1: Generate main.py only ───────────────────────────
print("🤖 Pass 1 — generating main.py...")
print("─" * 60)

API_PROMPT = """
Write ONLY the file main.py — a complete FastAPI Todo REST API.

Requirements:
- Pydantic models: TodoCreate (title: str), TodoUpdate (title: str | None, completed: bool | None), Todo (id, title, completed, created_at)
- In-memory dict storage, auto-incrementing int IDs
- Endpoints with correct status codes:
    GET    /todos          200 — list all
    POST   /todos          201 — create
    GET    /todos/{id}     200 or 404
    PUT    /todos/{id}     200 or 404
    DELETE /todos/{id}     204 or 404

Output a single fenced Python code block. First line of code inside the block must be:
# main.py
"""

r1, t1, n1 = generate(API_PROMPT, max_new_tokens=2048, temperature=0.1)
metrics.steps += 1
metrics.tokens_generated += n1
print(f"⚡ {n1:,} tokens in {t1:.1f}s")

# Validate it's actually a FastAPI app
if "FastAPI" not in r1 or "@app" not in r1:
    print("⚠️  Model didn't produce a FastAPI app — response preview:")
    print(r1[:600])
    raise RuntimeError("Generation failed: model ignored the task. Re-run this cell.")

blocks1 = extract_code_blocks(r1)
# Force correct filename regardless of what model used
api_code = next(iter(blocks1.values())) if blocks1 else ""
if not api_code:
    raise RuntimeError("No code block found in Pass 1 response.")

main_path = f"{OUTPUT_DIR}/main.py"
with open(main_path, "w") as f:
    f.write(api_code)
print(f"  📄 Written: {main_path}")

# ── PASS 2: Generate test_main.py given main.py ─────────────
print("\n🤖 Pass 2 — generating test_main.py...")
print("─" * 60)

TEST_PROMPT = f"""
Here is a FastAPI application in main.py:

```python
{api_code}
```

Write ONLY the file test_main.py — a pytest test suite for this API.

Requirements:
- Use pytest + httpx AsyncClient: `async with AsyncClient(transport=ASGITransport(app=app), base_url="http://test") as client:`
- Import: `from httpx import AsyncClient, ASGITransport`
- Import app: `from main import app`
- Mark every test: `@pytest.mark.anyio`
- At least 8 test functions covering: list todos, create todo, get by id, update, delete, 404 on missing id
- Fresh app state per test: reset the todos dict at the start of each test

Output a single fenced Python code block. First line inside must be:
# test_main.py
"""

r2, t2, n2 = generate(TEST_PROMPT, max_new_tokens=2500, temperature=0.1)
metrics.steps += 1
metrics.tokens_generated += n2
print(f"⚡ {n2:,} tokens in {t2:.1f}s")

if "def test_" not in r2:
    print("⚠️  Model didn't produce test functions — response preview:")
    print(r2[:600])
    raise RuntimeError("Generation failed: no test functions found. Re-run this cell.")

blocks2 = extract_code_blocks(r2)
test_code = next(iter(blocks2.values())) if blocks2 else ""
if not test_code:
    raise RuntimeError("No code block found in Pass 2 response.")

test_path = f"{OUTPUT_DIR}/test_main.py"
with open(test_path, "w") as f:
    f.write(test_code)
print(f"  📄 Written: {test_path}")
print(f"\n✅ Both files written. Proceed to run tests.")
````

````python
# Preview generated files
print("── main.py (first 30 lines) ──")
lines = open(f'{OUTPUT_DIR}/main.py').readlines()
print("".join(lines[:30]))
print(f"... ({len(lines)} lines total)\n")

print("── test_main.py (first 30 lines) ──")
lines = open(f'{OUTPUT_DIR}/test_main.py').readlines()
print("".join(lines[:30]))
print(f"... ({len(lines)} lines total)")
````

````python
# ── Step 3: Run the tests ────────────────────────────────────
print("\n🧪 Running pytest...\n")
passed, test_output = run_tests()

metrics.tests_passed = passed
metrics.total_time_s = time.time() - t_total_start

print(test_output)
print("─" * 60)
print(f"{'✅ Tests PASSED' if passed else '❌ Tests FAILED — intervention needed'}")
````

````python
# ── Step 4 (conditional): Self-correction loop ───────────────
MAX_RETRIES = 2

if not passed:
    print("\n📋 pytest output from initial run:")
    print(test_output)

    for attempt in range(MAX_RETRIES):
        metrics.interventions += 1
        print(f"\n🔄 Self-correction attempt {attempt + 1}/{MAX_RETRIES}...")

        # Read current files — truncate to stay within context budget
        def read_truncated(path, max_lines=60):
            if not os.path.exists(path):
                return "FILE NOT FOUND"
            lines = open(path).readlines()
            if len(lines) > max_lines:
                return "".join(lines[:max_lines]) + f"\n# ... ({len(lines)-max_lines} more lines truncated)"
            return "".join(lines)

        main_src  = read_truncated("/tmp/gemma_output/block_1.py")
        test_src  = read_truncated("/tmp/gemma_output/test_main.py")
        # Only send the last 800 chars of pytest output — the tail has the actual errors
        error_snippet = test_output.strip()[-800:]

        fix_prompt = f"""The pytest output below shows failures. Fix the code.

PYTEST ERRORS:
{error_snippet}

CURRENT main.py:
```python
{main_src}
```

CURRENT test_main.py:
```python
{test_src}
```

Rules:
- Output ONLY the two corrected files as fenced code blocks
- First block must start with `# main.py` on line 1
- Second block must start with `# test_main.py` on line 1
- No explanations, no extra text outside the code blocks
- All 8+ test functions must be present and passing
"""

        # More tokens for the fix — model needs room to write both files
        response, elapsed, n_tokens = generate(
            fix_prompt,
            max_new_tokens=3500,
            temperature=0.1,  # Lower temp for corrections — we want deterministic fixes
        )
        metrics.steps += 1
        metrics.tokens_generated += n_tokens
        print(f"   ⚡ {n_tokens:,} tokens in {elapsed:.1f}s")

        if n_tokens < 200:
            print("   ⚠️  Model output suspiciously short — it may have hit context limits")
            print(f"   Raw response: {repr(response[:300])}")

        blocks = extract_code_blocks(response)
        print(f"   📦 Extracted blocks: {list(blocks.keys())}")

        if not blocks:
            print("   ❌ No code blocks found in response — model didn't follow format")
            print(f"   Response preview: {response[:500]}")
            continue

        write_files(blocks)

        passed, test_output = run_tests()
        print(f"\n   pytest output:")
        print(test_output[-600:])

        if passed:
            metrics.errors_self_corrected += 1
            metrics.tests_passed = True
            print(f"   ✅ Fixed on attempt {attempt + 1}!")
            break
        else:
            print(f"   ❌ Still failing. {'Giving up.' if attempt == MAX_RETRIES - 1 else 'Retrying...'}")

metrics.total_time_s = time.time() - t_total_start
print(f"\n⏱️ Total elapsed: {metrics.total_time_s:.1f}s")
````
